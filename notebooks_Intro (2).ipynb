```python
"""
Student Grade Calculator
-------------------------
"""

from dataclasses import dataclass, field
from datetime import datetime

@dataclass
class Subject:
    name: str
    marks_obtained: float
    max_marks: float = 100

    @property
    def percentage(self) -> float:
        return (self.marks_obtained / self.max_marks) * 100

@dataclass
class Student:
    name: str
    subjects: list = field(default_factory=list)

    def add_subject(self, subject: Subject):
        self.subjects.append(subject)

    @property
    def total_percentage(self) -> float:
        if not self.subjects:
            return 0.0
        total_obtained = sum(s.marks_obtained for s in self.subjects)
        total_max = sum(s.max_marks for s in self.subjects)
        return (total_obtained / total_max) * 100

    @property
    def gpa(self) -> float:
        # Simple 4.0 scale conversion based on overall percentage
        pct = self.total_percentage
        if pct >= 90:
            return 4.0
        elif pct >= 80:
            return 3.7
        elif pct >= 70:
            return 3.3
        elif pct >= 60:
            return 3.0
        elif pct >= 50:
            return 2.5
        elif pct >= 40:
            return 2.0
        else:
            return 0.0
            
def get_letter_grade(percentage: float) -> str:
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    elif percentage >= 40:
        return "E"
    else:
        return "F (Fail)"


def get_float_input(prompt: str) -> float:
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("⚠️  Negative value is not allowed .")
                continue
            return value
        except ValueError:
            print("⚠️ write correct number.")


def get_int_input(prompt: str) -> int:
    while True:
        try:
            value = int(input(prompt))
            if value <= 0:
                print("⚠️  Value should be greater than 0.")
                continue
            return value
        except ValueError:
            print("⚠️  write correct number.")


def collect_student_data() -> Student:
    print("\n📘 Student Grade Calculator")
    print("-" * 35)
    name = input("Student's name: ").strip() or "Unknown Student"
    student = Student(name=name)

    num_subjects = get_int_input("number of subjects? ")

    for i in range(1, num_subjects + 1):
        print(f"\n--- Subject {i} ---")
        subject_name = input("Subject's name: ").strip() or f"Subject {i}"
        max_marks = get_float_input("Total marks (e.g. 100): ")
        obtained = get_float_input(f"Obtained marks (out of {max_marks}): ")
        while obtained > max_marks:
            print("⚠️  Obtained marks can not be greater than total marks.")
            obtained = get_float_input(f"Obtained marks (out of {max_marks}): ")
        student.add_subject(Subject(subject_name, obtained, max_marks))

    return student


def print_report(student: Student):
    print("\n" + "=" * 45)
    print(f"📊 RESULT CARD — {student.name}")
    print("=" * 45)
    print(f"{'Subject':<20}{'Marks':<15}{'Percentage':<10}")
    print("-" * 45)
    for s in student.subjects:
        marks_str = f"{s.marks_obtained}/{s.max_marks}"
        print(f"{s.name:<20}{marks_str:<15}{s.percentage:>6.2f}%")
    print("-" * 45)
    print(f"Overall Percentage : {student.total_percentage:.2f}%")
    print(f"Overall Grade      : {get_letter_grade(student.total_percentage)}")
    print(f"GPA (4.0 scale)    : {student.gpa:.2f}")
    print("=" * 45)


def save_report(student: Student):
    filename = f"{student.name.replace(' ', '_')}_report.txt"
    with open(filename, "w", encoding="utf-8") as f:
        f.write(f"RESULT CARD — {student.name}\n")
        f.write(f"Generated on: {datetime.now().strftime('%Y-%m-%d %H:%M')}\n")
        f.write("=" * 45 + "\n")
        f.write(f"{'Subject':<20}{'Marks':<15}{'Percentage':<10}\n")
        f.write("-" * 45 + "\n")
        for s in student.subjects:
            marks_str = f"{s.marks_obtained}/{s.max_marks}"
            f.write(f"{s.name:<20}{marks_str:<15}{s.percentage:>6.2f}%\n")
        f.write("-" * 45 + "\n")
        f.write(f"Overall Percentage : {student.total_percentage:.2f}%\n")
        f.write(f"Overall Grade      : {get_letter_grade(student.total_percentage)}\n")
        f.write(f"GPA (4.0 scale)    : {student.gpa:.2f}\n")
    print(f"\n✅ Report is save: {filename}")

if __name__ == "__main__":
    main()
```
